```python
def load_nc_file(file_path):
    ds = xr.open_dataset(file_path)
    
    var_name = list(ds.data_vars.keys())[0]
    # print(f"Variable name: {var_name}")
    data = ds[var_name].values

    ds.close()
    return data
```

```python
def build_yearly_array(nc_files):
    yearly_data = []
    for file in nc_files:
        data = load_nc_file(file)
        yearly_data.append(data)
    
    # yearly_data = np.array(yearly_data)
    # np.save('yearly_data.npy', yearly_data)
    return yearly_data
```

```python
input_feature_folder = 'ERA5_surface_solar_rad/'
nc_files = sorted([
    os.path.join(input_feature_folder, f)
    for f in os.listdir(input_feature_folder)
      if f.endswith('.nc')
]
)
yearly_data = build_yearly_array(nc_files)
print(f"Number of files processed: {len(yearly_data)}")

#make a new folder of npy files
output_folder = 'yearly_data_surface_solar_rad/'
os.makedirs(output_folder, exist_ok=True)
for i in range(len(yearly_data)):
    np.save(os.path.join(output_folder, f'yearly_data_{2001+i}.npy'), yearly_data[i])
```
